```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('housing.csv')

features = ['total_rooms', 'housing_median_age', 'total_bedrooms', 'median_income', 'median_house_value']
df = df[features].dropna()
df
```

```python
# Show the correlation matrix using a heatmap 
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix of Features")
plt.show()

```

```python
# Multicollinearity occurs when predictor variables are highly correlated with each other. 
# In the correlation heatmap above, if any pair of independent variables have a high correlation (e.g., > 0.8), 
# it can affect the stability and interpretability of linear regression models.
# This can be mitigated by removing one of the correlated features,
# using regularization (like Ridge/Lasso), or using techniques like PCA.
```
